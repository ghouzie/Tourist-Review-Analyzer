## Task A: 3 sentiment models



```python
sentiment_model_ids = [
    "distilbert-base-uncased-finetuned-sst-2-english",
    "cardiffnlp/twitter-roberta-base-sentiment-latest",
    "nlptown/bert-base-multilingual-uncased-sentiment",
]

sentiment_pipelines = {
    model_id: pipeline("sentiment-analysis", model=model_id)
    for model_id in sentiment_model_ids
}
```

## Inference


```python
def normalize_label(model_id: str, raw_label: str) -> str:
    """Map each model's own label scheme to a common positive/negative label."""
    raw_label = raw_label.upper()
    if model_id == "distilbert-base-uncased-finetuned-sst-2-english":
        return "positive" if raw_label == "POSITIVE" else "negative"
    if model_id == "cardiffnlp/twitter-roberta-base-sentiment-latest":
        return "positive" if raw_label == "POSITIVE" else "negative"  # NEUTRAL treated as negative
    if model_id == "nlptown/bert-base-multilingual-uncased-sentiment":
        stars = int(raw_label[0])
        return "positive" if stars >= 4 else "negative"
    raise ValueError(f"Unknown model_id {model_id}")


def predict_sentiment(text: str, model_id: str) -> dict:
    result = sentiment_pipelines[model_id](text)[0]
    return {
        "label": result["label"],
        "score": result["score"],
        "normalized_label": normalize_label(model_id, result["label"]),
        "metadata": "huggingface_AI_model",
    }


# quick sanity check on a few varied examples
for sample in [
    "The staff were amazing and the food was fresh!",
    "Cold food, rude staff, would not go back.",
    "Wow, another 45-minute wait for a table that was 'reserved'. Fantastic.",
]:
    print(sample)
    for model_id in sentiment_model_ids:
        print(" ", model_id, "->", predict_sentiment(sample, model_id))
    print()
```

## Evaluate Each Model

```python
results = {}
for model_id in sentiment_model_ids:
    preds = [predict_sentiment(t, model_id)["normalized_label"] for t in df["text"]]
    df[f"pred_{model_id}"] = preds
    acc = accuracy_score(df["true_label"], preds)
    f1 = f1_score(df["true_label"], preds, pos_label="positive")
    results[model_id] = {"accuracy": acc, "f1": f1}
    print(model_id)
    print("  accuracy:", round(acc, 3), " f1:", round(f1, 3))
    print("  confusion matrix [rows=true, cols=pred] (order: negative, positive):")
    print(" ", confusion_matrix(df["true_label"], preds, labels=["negative", "positive"]))
    print()

results_df = pd.DataFrame(results).T
results_df
```

## Zero-Shot Topic Classification



```python
topic_rows = [("The seafood platter at this place was incredibly fresh, probably the best I've had in Bahrain.", 'Food Quality'), ('Loved the view from the rooftop terrace, perfect spot for sunset dinner.', 'Atmosphere'), ('Staff went out of their way to accommodate our dietary restrictions. Excellent service.', 'Service'), ('Quick service, generous portions, and reasonable prices. Will definitely come back.', 'Price'), ("Best machboos I've tasted outside of a home kitchen.", 'Food Quality'), ('Cozy atmosphere, dim lighting, and a great playlist. Perfect date night spot.', 'Atmosphere'), ('The breakfast buffet was outstanding, so many options and everything was fresh.', 'Food Quality'), ('Friendly waiters and a beautiful garden seating area. Highly recommend.', 'Service'), ('Amazing value for money, the set menu had everything we needed.', 'Price'), ('The hotel room was spotless and the bed was so comfortable.', 'Cleanliness'), ('Their karak tea is unbeatable, I go there just for that.', 'Food Quality'), ('Great experience from start to finish, the manager even checked on us personally.', 'Service'), ('The dessert selection was incredible, especially the saffron pudding.', 'Food Quality'), ('Wonderful staff, they remembered our order from last time.', 'Service'), ('The pool area is gorgeous and well maintained, kids loved it.', 'Atmosphere'), ('Absolutely fantastic brunch, worth every dinar.', 'Price'), ('Efficient check-in process and a warm welcome from the front desk.', 'Service'), ('The grilled hammour was cooked to perfection.', 'Food Quality'), ('Spacious rooms with a great view of the sea, would stay again.', 'Location'), ('Delicious food, generous portions, and attentive staff throughout the meal.', 'Food Quality'), ('Waited over an hour for our food and it arrived cold.', 'Waiting Time'), ('The room smelled musty and the AC barely worked.', 'Cleanliness'), ('Rude staff at the front desk, made us feel unwelcome.', 'Service'), ('Overpriced for what you get, definitely not worth it.', 'Price'), ('Found a hair in my soup, will not be returning.', 'Cleanliness'), ("The place was filthy, tables weren't cleaned between customers.", 'Cleanliness'), ('Ordered medium steak, got something closer to shoe leather.', 'Food Quality'), ('Booked a sea view room, got a view of the parking lot instead.', 'Location'), ("Loud construction noise all night, couldn't sleep at all.", 'Atmosphere'), ('The manager was dismissive when we complained about the slow service.', 'Service'), ("Portion sizes were tiny for the price they're charging.", 'Price'), ("Bathroom wasn't cleaned properly, quite disappointing for the price.", 'Cleanliness'), ('Food tasted reheated, definitely not freshly prepared.', 'Food Quality'), ('Staff kept forgetting our order, had to remind them three times.', 'Service'), ('Air conditioning broke down mid-meal and nobody fixed it.', 'Atmosphere'), ('Terrible parking situation, had to walk 20 minutes in the heat.', 'Location')]

topic_df = pd.DataFrame(topic_rows, columns=["text", "true_topic"])
candidate_labels = ["Food Quality", "Service", "Price", "Cleanliness", "Atmosphere", "Location", "Waiting Time"]

zero_shot = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")

def predict_topic(text: str) -> dict:
    result = zero_shot(text, candidate_labels)
    return {
        "label": result["labels"][0],
        "score": result["scores"][0],
        "metadata": "huggingface_AI_model",
    }

topic_df["pred_topic"] = topic_df["text"].apply(lambda t: predict_topic(t)["label"])
topic_df.head()
```

```python
topic_acc = accuracy_score(topic_df["true_topic"], topic_df["pred_topic"])
topic_f1 = f1_score(topic_df["true_topic"], topic_df["pred_topic"], average="macro")
print("Zero-shot topic accuracy:", round(topic_acc, 3), " macro-F1:", round(topic_f1, 3))

topic_errors = topic_df[topic_df["true_topic"] != topic_df["pred_topic"]]
print(f"{len(topic_errors)} of {len(topic_df)} misclassified")
topic_errors
```

## Does the Model Understand My Domain?

---



- The sentiment models were trained on movie reviews, tweets, or generic product reviews, not Bahrain-specific restaurant/hotel reviews. Local dish names (machboos, karak, hammour) and mixed English/Arabic phrasing are outside their training distribution, which is a form of domain shift.
- The zero-shot model (`facebook/bart-large-mnli`) was trained for general natural-language-inference, not review classification specifically, it tends to do reasonably well on clear single-topic reviews but struggles when a review touches two topics at once (e.g., "cold food but great service").
 - None of the models have seen Arabic-language reviews, which real Bahrain review data would likely include.


Considering the accuracy/F1 figures mentioned above:
Use the pretrained model exactly as it is if accuracy is quite high (around 80%+) and failures are primarily on truly ambiguous or sarcastic language. Sarcasm identification is a challenging task even for humans.
Fine-tuning on local data would be worthwhile if errors concentrate around dialect/Arabic text or domain-specific terminology, as that's a systematic gap rather than random noise.
Improve preprocessing (separate reviews into sentences before categorizing) instead of fine-tuning if mixed-topic reviews for zero-shot are the cause of failures.
